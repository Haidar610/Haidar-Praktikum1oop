```python
#Latihan1
import numpy as np

# Membuat array harga
harga = np.array([5000, 7000, 3000, 12000, 4500])

# Menampilkan statistik dan perhitungan diskon
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Hasil [5000, 7000, 3000]  0.9 pada list biasa: Akan menghasilkan Error. kalau dikali desimal, Python tidak paham cara mengalikannya. Jika dikali angka bulat (misal * 2), list hanya akan diulang/direplikasi menjadi [5000, 7000, 3000, 5000, 7000, 3000], bukan dikalikan nilainya.
kenapa berbeda? karena NumPy Array didesain untuk operasi matematika langsung ke seluruh isi data sekaligus. kalau List standar Python didesain hanya untuk menyimpan kumpulan data, bukan angka.



```python
import pandas as pd

data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}

df = pd.DataFrame(data_kantin)
print(df)
```

kolom yang memiliki data kosong itu yang (None/NaN): Kolom menu (baris index 4) dan kolom terjual (baris index 2).
Risiko jika langsung dianalisis adalah Perhitungan matematika bisa menjadi tidak akurat atau salah hitung

```python
# Menampilkan 5 baris pertama
print(df.head())

# Informasi tipe data & jumlah data non-null
print(df.info())

# Statistik ringkas kolom numerik
print(df.describe())

# Ukuran dataset (baris, kolom)
print(df.shape)
```

Kolom dengan non-null lebih sedikit: Kolom menu dan terjual.
Artinya terdapat sel/baris yang bernilai kosong pada kolom tersebut, sehingga jumlah data yang terisi tidak sama dengan total baris keseluruhan.

```python
# Cek data kosong
print(df.isnull().sum())

# Isi kekosongan pada 'terjual' dengan angka 0
df['terjual'] = df['terjual'].fillna(0)

# Hapus baris jika kolom 'menu' kosong
df = df.dropna(subset=['menu'])

print(df)
```

Karena jika angka penjualan kosong, asumsi paling masuk akal adalah produk tersebut tidak ada yang terjual (0 unit).
Karena menu adalah identitas utama item. Jika nama menunya kosong/tidak diketahui, data harga dan angka terjual di baris tersebut tidak ada artinya

```python
# Cek duplikat
print("Jumlah duplikat:", df.duplicated().sum())

# Hapus duplikat
df = df.drop_duplicates()

# Ubah tipe data harga menjadi integer
df['harga'] = df['harga'].astype(int)

# Cek tipe data akhir
print(df.dtypes)
print(df)
```

1). Berkurang dari 4 baris menjadi 3 baris (karena baris Es Teh yang duplikat telah dihapus).
2). Jika kolom angka seperti harga terdeteksi sebagai teks (object), Pandas tidak akan bisa menghitung operasi matematika seperti total pendapatan atau rata-rata

```python
# 1. Filtering (terjual > 20)
laris = df[df['terjual'] > 20]
print("--- Menu Laris ---")
print(laris)

# 2. Sorting (urutkan penjualan terbanyak)
urut = df.sort_values(by='terjual', ascending=False)
print("\n--- Diurutkan ---")
print(urut)

# 3. Kolom Turunan (total_pendapatan)
df['total_pendapatan'] = df['harga'] * df['terjual']

# 4. Groupby (total pendapatan per menu)
ringkasan = df.groupby('menu')['total_pendapatan'].sum()
print("\n--- Total Pendapatan per Menu ---")
print(ringkasan)
```

1). Es Teh (menghasilkan total Rp300.0000)
2). Pengelola kantin bisa memperbanyak stok bahan baku Es Teh agar tidak kehabisan. Sebaliknya, untuk menu dengan pendapatan terendah/0 (seperti Mie Ayam), kantin bisa mengevaluasi rasa, memberi diskon, atau menggantinya dengan menu lain.
